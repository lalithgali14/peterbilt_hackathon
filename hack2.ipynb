```python
import pandas as pd
from itertools import combinations
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from mlxtend.frequent_patterns import apriori, association_rules

# Load datasets
option_code_info_df = pd.read_csv('Option Code Information file for UNT - Final(Sheet1).csv')
claim_info_df = pd.read_csv('Claim Information file for UNT(Sheet1).csv')

# Rename and merge
option_code_info_df.rename(columns={'Truck': 'Truck Number'}, inplace=True)
merged_df = pd.merge(option_code_info_df, claim_info_df, on='Truck Number', how='left')

# Ordinal Encoding for 'Scale Claim Cost' and 'Scale Labor Cost'
ordinal_mapping = {'Very Low': 1, 'Low': 2, 'Medium': 3, 'High': 4, 'Very High': 5}
merged_df['Scale Claim Cost'] = merged_df['Scale Claim Cost'].map(ordinal_mapping)
merged_df['Scale Labor Cost'] = merged_df['Scale Labor Cost'].map(ordinal_mapping)
merged_df['Scale Claim Cost'].fillna(0, inplace=True)  # Assuming 0 represents no claims

# Select features and target variable
X = merged_df.drop(columns=['Truck Number', 'Claim Number', 'Scale Claim Cost'])
y = merged_df['Scale Claim Cost']

# One-Hot Encode categorical variables
encoder = OneHotEncoder(sparse_output=False, drop='first')  # Updated parameter name
X_encoded = encoder.fit_transform(X)
X_encoded_df = pd.DataFrame(X_encoded, columns=encoder.get_feature_names_out(X.columns))


# Add pairwise interaction terms for Decision Tree model
for (col1, col2) in combinations(X_encoded_df.columns, 2):
    X_encoded_df[f"{col1}_x_{col2}"] = X_encoded_df[col1] * X_encoded_df[col2]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_encoded_df, y, test_size=0.3, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Instantiate and fit the Decision Tree model
dt_model = DecisionTreeClassifier(random_state=42, max_depth=10)  # Set max_depth as needed to prevent overfitting
dt_model.fit(X_train, y_train)

# Evaluate the Decision Tree
y_pred_dt = dt_model.predict(X_test)
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Classification Report:\n", classification_report(y_test, y_pred_dt))
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

# Create a binary DataFrame for the original one-hot encoded attributes without interaction terms
binary_df = X_encoded_df.copy()

# Filter data for high warranty costs (e.g., Scale Claim Cost >= 4)
high_cost_df = binary_df[y >= 4]

# Apply Apriori Algorithm to find frequent itemsets
frequent_itemsets = apriori(high_cost_df, min_support=0.01, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
rules = rules.sort_values(by="lift", ascending=False)

# Display top rules
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

# Plot the Decision Tree
plt.figure(figsize=(20, 10))
plot_tree(dt_model, feature_names=X_train.columns, filled=True, max_depth=3)  # Set max_depth for readability
plt.title("Decision Tree Visualization")
plt.show()
```

```python
# Get feature importance from Decision Tree
importances = dt_model.feature_importances_
features = X_train.columns

# Sort and plot the top features by importance
feature_importance_df = pd.DataFrame({'Feature': features, 'Importance': importances}).sort_values(by='Importance', ascending=False)

plt.figure(figsize=(12, 8))
plt.barh(feature_importance_df['Feature'][:20], feature_importance_df['Importance'][:20])
plt.gca().invert_yaxis()
plt.xlabel('Feature Importance')
plt.ylabel('Features')
plt.title('Top 20 Features by Importance in Decision Tree Model')
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Generate predictions and confusion matrix
y_pred_dt = dt_model.predict(X_test)
cm = confusion_matrix(y_test, y_pred_dt)

# Plot confusion matrix
plt.figure(figsize=(8, 6))
ConfusionMatrixDisplay(cm).plot(cmap="Blues")
plt.title("Confusion Matrix for Decision Tree Model")
plt.show()
```

```python
import networkx as nx

# Convert association rules to a network
rules_filtered = rules[(rules['lift'] > 1.2) & (rules['confidence'] > 0.7)]
G = nx.Graph()

for idx, row in rules_filtered.iterrows():
    for ant in row['antecedents']:
        for con in row['consequents']:
            G.add_edge(ant, con, weight=row['lift'])

plt.figure(figsize=(12, 12))
pos = nx.spring_layout(G, k=0.5)
nx.draw(G, pos, with_labels=True, node_size=3000, node_color="skyblue", font_size=12, font_weight="bold", edge_color="gray")
plt.title("Association Rules Network for High-Cost Attribute Pairs")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Filter the association rules based on confidence and lift thresholds
rules_filtered = rules[(rules['lift'] > 1.2) & (rules['confidence'] > 0.7)]

# Initialize a NetworkX graph
G = nx.Graph()

# Add nodes and edges based on association rules
for idx, row in rules_filtered.iterrows():
    # Each rule has antecedents (LHS) and consequents (RHS)
    for antecedent in row['antecedents']:
        for consequent in row['consequents']:
            # Add an edge with the 'lift' value as the weight
            G.add_edge(antecedent, consequent, weight=row['lift'])

# Draw the graph
plt.figure(figsize=(12, 12))
pos = nx.spring_layout(G, k=0.5)  # Positioning of nodes
nx.draw(
    G, pos, with_labels=True, 
    node_size=3000, node_color="skyblue", 
    font_size=12, font_weight="bold", edge_color="gray"
)

# Optionally add edge labels for lift values
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Network of Association Rules for High-Cost Attribute Pairs")
plt.show()
```
